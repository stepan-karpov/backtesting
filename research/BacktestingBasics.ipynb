```python
from tools import ROOT
from tools.readers.lighter import LobReaderRaw, LobReader, TradesReaderRaw, TradesReader, TickerReaderRaw, TickerReader, MarketStatsReaderRaw, MarketStatsReader
```

```python
lob_paths = [
    ROOT / f"data/lighter/DOGE/lob/202607{day:02d}/{hour:02d}.parquet"
    for day in range(12, 13)
    for hour in range(0, 10)
]
trades_paths = [
    ROOT / f"data/lighter/DOGE/trades/202607{day:02d}/{hour:02d}.jsonl.zst"
    for day in range(12, 13)
    for hour in range(0, 10)
]
tickers_paths = [
    ROOT / f"data/lighter/DOGE/ticker/202607{day:02d}/{hour:02d}.jsonl.zst"
    for day in range(12, 13)
    for hour in range(0, 10)
]
market_stats_paths = [
    ROOT / f"data/lighter/DOGE/market_stats/202607{day:02d}/{hour:02d}.jsonl.zst"
    for day in range(12, 13)
    for hour in range(0, 10)
]
```

```python
lobs = LobReader(lob_paths).load()

lobs
```
